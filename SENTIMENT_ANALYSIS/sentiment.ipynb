# 2. Data Preprocessing

### 2.1 •	Removing special characters, numbers, and HTML tags

```python
import re
```

```python
def clean_text(text):
    text = str(text)                            # Convert to string
    text = text.lower()                         # Lowercase
    text = re.sub(r'<.*?>', '', text)           # Remove HTML tags
    text = re.sub(r'[^a-zA-Z ]', ' ', text)     # Remove numbers & special characters
    text = re.sub(r'\s+', ' ', text)           # Remove extra whitespaces
    return text.strip()
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    if text is None:
        return ""
    txt = str(text).lower()
    try:
        tokens = word_tokenize(txt)
    except LookupError:
        # fallback simple tokenizer (alphabetic tokens only)
        tokens = re.findall(r'\b[a-z]+\b', txt)
    filtered = [w for w in tokens if w.isalpha() and w not in stop_words]
    return ' '.join(filtered)
```

### 2.6 lemmatization

```python
# ...existing code...
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
import re

# ensure wordnet (and punkt) are available (download to ./nltk_data if missing)
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet', download_dir='./nltk_data')
    nltk.download('omw-1.4', download_dir='./nltk_data')
    nltk.data.path.append('./nltk_data')

try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt', download_dir='./nltk_data')
    nltk.data.path.append('./nltk_data')

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

def safe_tokenize(text):
    if text is None:
        return []
    txt = str(text).lower()
    try:
        tokens = word_tokenize(txt)
    except LookupError:
        tokens = re.findall(r'\b[a-z]+\b', txt)
    return [t for t in tokens if t.isalpha()]

def stem_words(tokens):
    return [stemmer.stem(word) for word in tokens]

def lemmatize_stemmed(tokens):
    # tokens expected to be a list of stemmed tokens; handle other cases robustly
    if not isinstance(tokens, (list, tuple)):
        return []
    out = []
    for w in tokens:
        try:
            out.append(lemmatizer.lemmatize(str(w), pos='v'))
        except Exception:
            out.append(w)
    return out

# ensure Tweet_stemmed exists — create if necessary
if 'Tweet_stemmed' not in data.columns:
    if 'Tweet_tokens' in data.columns:
        data['Tweet_stemmed'] = data['Tweet_tokens'].apply(stem_words)
    else:
        data['Tweet_tokens'] = data['Tweet'].apply(safe_tokenize)
        data['Tweet_stemmed'] = data['Tweet_tokens'].apply(stem_words)

# Now lemmatize the stemmed tokens
data['Tweet_lemmatized'] = data['Tweet_stemmed'].apply(lemmatize_stemmed)
# ...existing code...
```

### 3.3.1 Word2Vec Embedding with Gensim

```python
from gensim.models import Word2Vec

# data['Tweet_lemmatized'] should be a list of tokens for each tweet
w2v_model = Word2Vec(
    sentences=data['Tweet_lemmatized'],
    vector_size=100,   # size of embedding vectors
    window=5,          # context window
    min_count=1,       # ignore words with low frequency
    workers=4,         # number of CPU threads
    sg=1               # 1: skip-gram, 0: CBOW
)
```

```python
import numpy as np
word = 'murder'
if word in w2v_model.wv:
    vector = w2v_model.wv[word]
else:
    vector = np.zeros(w2v_model.vector_size)

print(vector)  # Get the vector for the word 'happy'
print(vector.shape)
```

```python
import numpy as np

def tweet_vector(tokens):
    tokens = [token for token in tokens if token in w2v_model.wv]
    if not tokens:
        return np.zeros(w2v_model.vector_size)
    return np.mean(w2v_model.wv[tokens], axis=0)

data['Tweet_w2v'] = data['Tweet_lemmatized'].apply(tweet_vector)
```

```python
from gensim.models import KeyedVectors

glove_model = KeyedVectors.load_word2vec_format(word2vec_output_file, binary=False)
```

```python
print(glove_model['happy'])  # 300-dimensional vector for 'happy'
```

```python
import numpy as np
def tweet_glove_vector(tokens):
    tokens = [token for token in tokens if token in glove_model]
    if not tokens:
        return np.zeros(glove_model.vector_size)
    return np.mean([glove_model[token] for token in tokens], axis=0)

data['Tweet_glove'] = data['Tweet_lemmatized'].apply(tweet_glove_vector)
```

### 3.4 •	Pre-trained language models (BERT).

```python
import numpy as np
import torch
from transformers import BertTokenizer, BertModel

# 1) Load BERT
bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
bert_model = BertModel.from_pretrained("bert-base-uncased")
bert_model.eval()

# 2) Embedding function
def get_bert_embeddings(text):
    inputs = bert_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )
    with torch.no_grad():
        outputs = bert_model(**inputs)
        cls_embedding = outputs.last_hidden_state[:, 0, :].squeeze().numpy()
    return cls_embedding

# 3) Apply on your cleaned text
data["Tweet_bert"] = data["Tweet_lem_text"].apply(get_bert_embeddings)

# 4) Feature + label
X_bert = np.stack(data["Tweet_bert"].values)
y = data["Sentiment"]
```

```python
from transformers import AutoTokenizer, AutoModel

bt_tokenizer = AutoTokenizer.from_pretrained("vinai/bertweet-base")
bt_model = AutoModel.from_pretrained("vinai/bertweet-base")
bt_model.eval()

def get_bertweet_embeddings(text):
    inputs = bt_tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )
    with torch.no_grad():
        outputs = bt_model(**inputs)
        cls_embedding = outputs.last_hidden_state[:, 0, :].squeeze().numpy()
    return cls_embedding

data["Tweet_bertweet"] = data["Tweet_lem_text"].apply(get_bertweet_embeddings)
X_bertweet = np.stack(data["Tweet_bertweet"].values)
```

# Model selection

### Importing necessary library

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score
```

### Train Test Split

```python

X_train, X_test, y_train, y_test = train_test_split(X_bertweet, y, test_size=0.2, random_state=42)
```

### Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Logistic regression Model

```python
from sklearn.linear_model import LogisticRegression
lr_bt = LogisticRegression(max_iter=2000, random_state=42)
lr_bt.fit(X_train, y_train)
```

### Logistic regression report

```python
y_pred = lr_bt.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred, average='weighted'))
```

# SVM Model By GridSearchCV 

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
param_grid = {'C': [0.1, 1, 10], 'gamma': ['scale', 0.01, 0.1]}
grid = GridSearchCV(SVC(), param_grid, cv=3, scoring='f1_weighted')
grid.fit(X_train_scaled, y_train)
```

### SVM Model Report(GridSearchCV)

```python
# After GridSearchCV fits, get best F1:
y_pred_best = grid.best_estimator_.predict(X_test_scaled)

from sklearn.metrics import f1_score, classification_report

best_f1 = f1_score(y_test, y_pred_best, average='weighted')
print("Tuned SVM F1:", best_f1)
print(classification_report(y_test, y_pred_best))


```

```python
# Prediction using best SVM from GridSearchCV
y_pred_best = grid.best_estimator_.predict(X_test_scaled)

from sklearn.metrics import f1_score, classification_report, accuracy_score

# F1-score (weighted)
best_f1 = f1_score(y_test, y_pred_best, average='weighted')

# Accuracy (optional but useful)
best_acc = accuracy_score(y_test, y_pred_best)

print("Tuned SVM Accuracy:", best_acc)
print("Tuned SVM F1-score (weighted):", best_f1)
print("\nTuned SVM Classification Report:\n")
print(classification_report(y_test, y_pred_best))
```
